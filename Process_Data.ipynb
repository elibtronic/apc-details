## Dataset Creation

loads up Google Sheet, processes all the columns
puts it somewhere it can get scooped up by GH and the APP

```python
def make_dataset(JOURNAL_URL, PUB_URL):
    journalDF = pd.read_csv(JOURNAL_URL)
    pubDF = pd.read_csv(PUB_URL)
    combinedDF = pd.merge(journalDF,pubDF, on="Publisher")
    return combinedDF
```
