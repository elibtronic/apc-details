# Exploratory ALMA analysis

This will look through the export from Alma in attempt to make a standardized list of titles.

```python
import pandas as pd
import sqlite3
```

```python
alma_df = pd.read_csv("ALMA_export.csv")
#alma_df.sample(10)
#len(alma_df)
```

```python
#Build the list of publisher names
pub_names = [
    "American Chemical Society",
    "Association for Computing Machinery",
    "Cambridge University Press",
    "Canadian Science Publishing",
    "JMIR",
    "MDPI",
    "Microbiology Society Journals",
    "Oxford University Press",
    "Royal Society of Chemistry",
    "Francis",
    "Wiley",
    ]
```

```python
#Search, rename
selected_pub_df = pd.DataFrame()

for pub in pub_names:
    
    sub_pub = alma_df[alma_df["Publisher"].str.contains(pub, regex=True, na=False)]
    sub_pub["Publisher"] = pub
    
    selected_pub_df = pd.concat([selected_pub_df,sub_pub])
```
